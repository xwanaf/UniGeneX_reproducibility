```python
import logging
import sys

def configure_logging(logger_name):
    LOG_LEVEL = logging.DEBUG
    log_filename = logger_name+'.log'
    importer_logger = logging.getLogger('importer_logger')
    importer_logger.setLevel(LOG_LEVEL)
    formatter = logging.Formatter('%(asctime)s : %(levelname)s : %(message)s')

    fh = logging.FileHandler(filename=log_filename)
    fh.setLevel(LOG_LEVEL)
    fh.setFormatter(formatter)
    importer_logger.addHandler(fh)

    sh = logging.StreamHandler(sys.stdout)
    sh.setLevel(LOG_LEVEL)
    sh.setFormatter(formatter)
    importer_logger.addHandler(sh)
    return importer_logger
```

## Overall

```python
from matplotlib import cm

adata_vis = sp_log
fig, axs = plt.subplots(1,1,figsize = (21, 14))
he_spatial = sc.pl.spatial(
    adata_vis, 
    color=None, 
    ax = axs, 
    alpha_img = .9, 
#     palette = color_dict_use,#sp_log.uns[f'{method_name}_colordict'],
    crop_coord = None, #[x_start / sf, x_end / sf, y_start / sf, y_end / sf],
    alpha = .0,
    spot_size = 10,
    return_fig = True,
    show = False,
)

adata_input = adata_sp_time.copy()

float_values = adata_sp_time.obs['time']
lower, upper = np.quantile(float_values, .05), np.quantile(float_values, .95)
adata_input.obs['time'] = adata_sp_time.obs['time'].clip(lower=lower, upper=upper)
adata_input = adata_input[adata_sp_time.obs['sample'] == sample]


x_centroid = np.array(adata_input.obs['x_centroid'] * sf)
y_centroid = np.array(adata_input.obs['y_centroid'] * sf)
height = np.array(adata_input.obs['time'])

dist_mask_r = 5e4

exp_dist_std = 1e5
x = np.linspace(min(x_centroid), max(x_centroid), 500)
y = np.linspace(min(y_centroid), max(y_centroid), 500)
X, Y = np.meshgrid(x, y)

# Interpolate heights at the meshgrid points
Z = np.zeros_like(X)
for x_c, y_c, h in zip(x_centroid, y_centroid, height):
    
    dist_square = ((X - x_c)**2 + (Y - y_c)**2)
    exp_dist = np.exp(-dist_square / (exp_dist_std * sf))
    indicator = (dist_square < dist_mask_r).astype(int)
    Z += h * indicator * np.exp(-((X - x_c)**2 + (Y - y_c)**2) / (exp_dist_std * sf))  # 5e4

Z_out = Z.copy()
Z_dist = np.zeros_like(X)
for x_c, y_c, h in zip(x_centroid, y_centroid, height):
    
    dist_square = ((X - x_c)**2 + (Y - y_c)**2)
    exp_dist = np.exp(-dist_square / (exp_dist_std * sf))
    indicator = (dist_square < dist_mask_r).astype(int)
    Z_dist += indicator * np.exp(-((X - x_c)**2 + (Y - y_c)**2) / (exp_dist_std * sf))  # 5e4

index = Z_dist > 0
Z[index] = Z[index] / Z_dist[index]
# Z[~index] = 0

levels = np.linspace(Z[Z > 0].min(), Z.reshape(-1).max() + 0.01, 12)
if Z[Z > 0].min() > 0.01:
    levels = [0.01] + levels.tolist()
else:
    pass


cmap = cm.get_cmap('Spectral').reversed()  
norm = plt.Normalize(vmin=levels[2], vmax=levels[-1])
colors = cmap(norm(levels))

contour = axs.contourf(
    X, Y, Z, 
    levels = levels, 
    linewidths = 1., 
#     cmap = cmap, #vlag
#     vmin = levels[0], vcenter = 3,
    colors = colors,
    alpha = 0.6, 
    zorder = 40
)
cb = plt.colorbar(contour, pad = 0.01, format = '%.2f')
cb.set_ticks(np.linspace(levels[0], levels[-1], num=10))  
cb.update_ticks()  # Update the ticks

axs.clabel(contour, inline=False, fontsize=8, zorder = 50, colors = 'black', fmt='%1.2f')


plt.show()
plt.close()
         
```
