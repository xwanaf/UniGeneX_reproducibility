```python
# build large-scale data in scBank format from a group of AnnData objects
# %%
import gc
import json
from pathlib import Path
import argparse
import shutil
import traceback
from typing import Dict, List, Optional
import warnings
import numpy as np
import os
import anndata

import scanpy as sc
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


from typing import Any, Dict, List, Mapping, Optional, Tuple, Union
from typing_extensions import Self, Literal
from scipy.sparse import issparse

import matplotlib as mpl
def set_vis_default(font = 1.6, circle_marker = 1.):
    mpl.rcParams['font.family'] = 'Arial'
    plt.rcParams.update(plt.rcParamsDefault)
#     %matplotlib inline
    try:
        plt.style.use('seaborn-whitegrid')
    except:
        plt.style.use('seaborn-v0_8-whitegrid')

    sns.set_context('paper',font_scale=font)
    plt.rcParams["legend.markerscale"] = circle_marker
    mpl.rcParams['font.family'] = 'Arial'

def get_color_dict(adata, col, uns_key):
    c_dict = dict(zip(adata.obs[col].astype('category').unique().categories, adata.uns[uns_key]))
    return c_dict
```

```python
data_paths = []

# ===================================
base_path = '/import/home2/xwanaf/Img2Expr/data/lung_HLCA/pretrain_v4'
data_path = Path(base_path) / 'leiden_1e3_epi'
print(os.listdir(data_path))
data_file_names = [
    'leiden_2p5_epi_v1.csv', 'leiden_2p5.csv', 'leiden_2p5_epi_v2.csv', 'leiden_2p5_epi_v3.csv', 'leiden_2p5_epi_v7.csv', 
    'leiden_2p5_epi_v4.csv', 'leiden_1p_fibro_v1.csv'
]
data_paths = data_paths + [data_path / data_file_name for data_file_name in data_file_names]


for data_path in data_paths:
    add_col = Path(data_path).stem
    if add_col == 'leiden_2p5':
        add_col = 'leiden_2p5_epi'
    cell_type_refine_leiden_1p_df = pd.read_csv(data_path, index_col = 0)
    if cell_type_refine_leiden_1p_df.shape[0] < atlas_adata.shape[0]:
        tmp = np.empty(atlas_adata.shape[0], dtype = object)
        index = atlas_adata.obs.index.isin(cell_type_refine_leiden_1p_df.index)
        tmp[index] = cell_type_refine_leiden_1p_df.loc[atlas_adata[index].obs.index][cell_type_refine_leiden_1p_df.columns[0]]

        atlas_adata.obs[add_col] = tmp
        atlas_adata.obs[add_col] = atlas_adata.obs[add_col].astype('category')


    else:
        cell_type_refine_leiden_1p_df = cell_type_refine_leiden_1p_df.loc[atlas_adata.obs.index]
#         assert np.array_equal(cell_type_refine_leiden_1p_df.index, atlas_adata.obs.index)


        add_col = Path(data_path).stem
        atlas_adata.obs[add_col] = cell_type_refine_leiden_1p_df[cell_type_refine_leiden_1p_df.columns[0]].copy()
        atlas_adata.obs[add_col] = atlas_adata.obs[add_col].astype('category')
```

```python
import matplotlib.patheffects as path_effects
set_vis_default(font = .8, circle_marker = 3.)

# ct_all = adata_all.obs['dataset_names'].astype('category').unique().categories

adata_vis = atlas_adata
color_dict = get_color_dict(atlas_adata_ref, col = 'ct_v7', uns_key = 'ct_v7_colors')

fig, axs = plt.subplots(1, 2, figsize=(24, 12), dpi = 150, gridspec_kw={'wspace': .2, 'hspace': 0.05})
# color_dict = get_color_dict(atlas_adata, col = 'ct_v7', uns_key = 'ct_v7_colors')

sc.pl.umap(
    adata_vis, color='ct_vis', 
    size=1, frameon=False, show=False, ax=axs[0], 
    palette = color_dict, 
#     groups = groups#,
    legend_loc = None #'on data' 
)
# axs[0].legend(bbox_to_anchor=(1., 1.0))
axs[0].set_title('')

# Assuming adata_vis is your AnnData object and color_dict is defined

label_cts = [
    'NK cells',
 'CD4 T cells',
 'CD8 T cells',
 'Mast cells',
 'B cells',
 'Plasma cells',
 'Plasmacytoid DCs',
    
    'Lymphatic EC mature',
 'EC arterial',
 'EC venous systemic',
 'EC aerocyte capillary',
 'EC general capillary',
    
    'Pericytes',
 'SMCs',
    
]

indices_to_label = adata_vis.obs.index[adata_vis.obs['ct_vis'].isin(label_cts)].tolist()

# Loop through the indices and add labels
for ct in label_cts:
    index = adata_vis.obs['ct_vis'].isin([ct])
    # Get the UMAP coordinates
    x = adata_vis.obsm['X_umap'][index, 0]
    y = adata_vis.obsm['X_umap'][index, 1]
    x_mid, y_mid = np.median(x), np.median(y)

    # Add label (replace 'Label' with your desired label)
    axs[0].text(
        x_mid, y_mid, ct, fontsize=15, 
        ha='center', va='center', 
        fontweight = 'bold', 
        path_effects=[path_effects.withStroke(linewidth=1, foreground='white')]
    )

axs[1].remove()


plt.show()
plt.close()
```
